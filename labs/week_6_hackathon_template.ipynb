```text
project-name/
  README.md           # team members, problem, installation and how to run
  requirements.txt    # or another environment specification
  src/                # application and loop code, if using modules
  notebooks/          # experiments or the main runnable notebook
  data/               # shareable sample data or download instructions
  reports/            # comparison results, loop traces and presentation
```

## 3) Design the loop

Sketch the path from the user's task to the checked result:

**Task → model → tool execution → observed result → check → revision or stop**

Show:

- the data and context sent to each model call;
- the tools used and the results returned;
- what triggers revision, additional retrieval or clarification, if relevant;
- the attempt/time limit and the condition that stops the loop;
- the independent check used to evaluate the final result.

Use the components your problem needs: generated Python, simulation, classification or retrieval. Explain your choice of model and baseline. If the workflow asks a user for clarification, wait for their actual reply.

## 5) Evaluate against a baseline

Choose a simple baseline and use the same held-out tasks or data for both systems. Define the expected results before inspecting the final outputs.

Record:

- **Quality:** a task-appropriate measure, such as classification precision/recall, numerical error, verified answer facts or simulation checks.
- **Time:** total elapsed time, including review and repair. Identify any time spent waiting for a person.
- **Model usage:** input/output tokens and API cost where available. Leave unavailable usage unknown.
- **Loop behaviour:** a trace showing inputs, actions, observations, revisions and the stopping condition. Show a recorded failure and how it was handled.

Include a normal case and a difficult case, such as missing data, an ambiguous request or a failed check. Report the actual outcome, including unresolved failures. Explain whether the loop improved the result enough to justify its extra work.
